## Part 2: PyTorch Tensor Fundamentals
Tensors are N-dimensional arrays designed for GPU computing.

```python
import torch

# 1. Creating Tensors
x = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
y = torch.ones((2, 2), dtype=torch.float32)

print("Tensor X:\n", x)
print("Shape of X:", x.shape)
print("Data Type:", x.dtype)

# 2. Tensor Operations (Matrix Multiplication)
z = torch.matmul(x, y)
print("\nMatrix Multiplication Result (z):\n", z)

# 3. Moving Tensors to GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
x_gpu = x.to(device)
print(f"\nTensor execution device: {x_gpu.device}")
```

## Part 3: Autograd (Automatic Differentiation)
PyTorch tracks operations on tensors marked with `requires_grad=True` to compute gradients via backpropagation.

```python
# Initialize trainable parameters (weight and bias)
w = torch.tensor([2.0], requires_grad=True)
x = torch.tensor([3.0])
b = torch.tensor([1.0], requires_grad=True)

# Forward pass: linear equation y = w * x + b
y = w * x + b  # Expected: 2 * 3 + 1 = 7.0

# Compute gradients automatically
y.backward()

# Print gradients dy/dw and dy/db
print("y =", y.item())
print("Gradient dy/dw (should equal x = 3.0):", w.grad.item())
print("Gradient dy/db (should equal 1.0):", b.grad.item())
```
