**Strategy 3 - Statistical Imputation (Mean vs. Median)**

Business Context: We are missing 177 ages. We can't drop them (too much data loss), so we fill them.

Mean: Good for "average" passengers.

Median: Better if there are extreme outliers (like a few 80-year-olds in a sea of 20-somethings).

```python
import pandas as pd

# Loading data from a URL or local file
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# High-level overview
# print(df.info()) # Shows data types and non-null counts
```

```python
# Creating a copy to compare
df_mean_fill = df.copy()
df_median_fill = df.copy()
```

```python

mean_value = df_mean_fill['Age'].mean()
median_value = df_median_fill['Age'].median()

# Filling Age
df_mean_fill['Age'] = df_mean_fill['Age'].fillna(mean_value)
df_median_fill['Age'] = df_median_fill['Age'].fillna(median_value)

print(f"Mean Age: {df_mean_fill['Age'].mean():.2f}")
print(f"Median Age: {df_median_fill['Age'].median():.2f}")
```

**Strategy 4 - Group-Based Imputation (The "Pro" Way)**

Business Context: A 1st Class passenger is likely to be older than a 3rd Class passenger. Instead of giving everyone the global average age, let's give them the average age of their Ticket Class.

```python
import pandas as pd

# Loading data from a URL or local file
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# High-level overview
# print(df.info()) # Shows data types and non-null counts
```

```python
df_groupbased = df.copy()
```

```python
# Fill missing Age based on the average age of their Pclass (1, 2, or 3)
median_transform = df_groupbased.groupby('Pclass')['Age'].transform('median')

df_groupbased['Age'] = df_groupbased['Age'].fillna(median_transform)
```

```python

print("Check: Are there any missing ages left?")
print(df_groupbased['Age'].isnull().sum())
```

```python
print(f"Median Age: {df_median_fill['Age'].median():.2f}")
print(f"Median Age Groupby: {df_groupbased['Age'].median():.2f}")
```
